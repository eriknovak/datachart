# Themes

This section showcases the themes found in the [datachart.themes](/references/themes/) module and how to customize them. Seven predefined themes are available: `DEFAULT`, `GREYSCALE`, `PUBLICATION`, `BACKGROUND`, `MINIMAL`, `MATERIAL`, and `ACADEMIC` — see the [Theme Gallery](theme-gallery.ipynb) for every theme rendered across the full range of chart types.

Themes may also carry defaults for chart settings ([ThemeDefaultAttrs](/references/typings/#datachart.typings.ThemeDefaultAttrs)): `chart_default_show_grid` supplies the grid when a chart call leaves `show_grid` unset (every predefined theme ships a muted `"y"` grid), `chart_default_show_values` does the same for bar value labels (on in `MINIMAL`, `MATERIAL`, and `ACADEMIC`), and `plot_hatch_cycle` assigns hatch patterns per bar/histogram series (only `ACADEMIC` ships one). An explicit chart setting always wins over the theme default.

Let's start by importing the necessary functions to help us work with the `datachart.themes` module.

```python
import random
import numpy as np
from datachart.charts import (
    BarChart,
    BoxPlot,
    Heatmap,
    Histogram,
    LineChart,
    ParallelCoords,
    ScatterChart,
)
from datachart.constants import VALFMT, FIG_SIZE, LINE_STYLE, SHOW_GRID
```

```python
from datachart.config import config
```

To get the supported themes, you have to load them from the `datachart.themes` module.

```python
from datachart.constants import THEME
```

The [datachart.constants.THEME](/references/constants/#datachart.constants.THEME) module contains all the predefined themes.

## Applying a Theme

Applying a theme replaces the whole global configuration, so set it before building the charts it should style:

```python
config.set_theme(THEME.MINIMAL)

_ = BarChart(
    data=[{"label": f"cat{idx}", "y": 10 + 5 * idx} for idx in range(5)],
    title="Bar chart under THEME.MINIMAL",
    figsize=FIG_SIZE.A4_NARROW,
)
```

To return to the default theme, reset the configuration:

```python
config.reset_config()
```

See the [Theme Gallery](theme-gallery.ipynb) for every predefined theme rendered across the full range of chart types.

Once you define the theme, you can use it by updating the `config` module in the following way:

```python
from datachart.config import config
```

```python
config.update_config(CUSTOM_THEME)
```

Once you do this, all the plots will use the custom theme.

**Bar Chart**

```python
_ = BarChart(
    data=[
        {"label": f"xx{id}", "y": 100 * (id + 1) * random.random()}
        for id in range(10)
    ],
    vlines=[{"x": 2 * i} for i in range(1, 4)],
    hlines={"y": 400},
    title="Title",
    xlabel="the global x-axis label",
    ylabel="the global y-axis label",
    figsize=FIG_SIZE.A4_NARROW,
    show_grid=SHOW_GRID.BOTH,
    xmin=-0.5,
    xmax=9.5,
)
```

**Line Chart**

```python
_ = LineChart(
    data=[
        [{"x": x / 10, "y": np.cos(x / 2)} for x in range(21)],
        [{"x": x / 10, "y": np.sin(x / 2)} for x in range(21)],
    ],
    subtitle=["cosine", "sine"],
    title="Title",
    xlabel="the global x-axis label",
    ylabel="the global y-axis label",
    figsize=FIG_SIZE.A4_NARROW,
    show_grid=SHOW_GRID.BOTH,
    show_legend=True,
)
```

**Scatter Chart**

```python
chart_data_bubble_hue = [
    {
        "x": random.uniform(0, 10),
        "y": random.uniform(0, 10),
        "population": random.uniform(100, 1000),
        "region": random.choice(["North", "South", "East", "West"])
    }
    for _ in range(50)
]
```

```python
_ = ScatterChart(
    data=chart_data_bubble_hue,
    size="population",
    hue="region",
    size_range=(30, 250),
    title="Title",
    xlabel="the global x-axis label",
    ylabel="the global y-axis label",
    figsize=FIG_SIZE.A4_NARROW,
    show_grid=SHOW_GRID.BOTH,
    show_legend=True,
)
```

Finally, reset the configuration back to the default theme:

```python
config.reset_config()
```
